```python
import pandas as pd

# Caminho para o arquivo CSV
caminho_arquivo = '2026.csv'

print("A carregar os dados...")
# sep=None e engine='python' tentam detetar automaticamente se o separador é ';' ou ','
df = pd.read_csv(caminho_arquivo, sep=None, engine='python')

# 1. Ver as primeiras 5 linhas para conhecer as colunas
print("\n--- Primeiras Linhas ---")
display(df.head())

# 2. Ver o nome de todas as colunas e os tipos de dados
print("\n--- Informações Estruturais ---")
df.info()

# 3. Verificar se existem valores nulos
print("\n--- Valores Nulos por Coluna ---")
print(df.isnull().sum())
```
